```python
orders_file1 = pd.read_sql("SELECT * FROM orders", engine)
order_items_file2= pd.read_sql("SELECT * FROM order_items", engine)
```

```python
rfm = pd.merge(orders_file1, order_items_file2, on="order_id")
```

```python
rfm["order_date"] = pd.to_datetime(rfm["order_date"])
```

```python
today = pd.Timestamp.today()
```

```python
last_order = rfm.groupby("customer_id")["order_date"].max()
recency = (today - last_order).dt.days
```

```python
frequency = rfm.groupby("customer_id")["order_id"].count()
```

```python
monetary = rfm.groupby("customer_id")["total_price"].sum()
```

```python
rfm_data = pd.DataFrame({
    "Recency": recency,
    "Frequency": frequency,
    "Monetary": monetary
})
```

```python
print(rfm_data)
```

```python
def segment(row):
    if row['Monetary'] > 5000:
        return "High Value"
    elif row['Monetary'] > 2000:
        return "Medium Value"
    else:
        return "Low Value"

rfm_data['Segment'] = rfm_data.apply(segment, axis=1)
```

```python
rfm_data.to_sql(
    name="customer_segments",
    con=engine,
    if_exists="replace",
    index=False
)
```

```python
rfm_data = rfm_data.reset_index()
```
